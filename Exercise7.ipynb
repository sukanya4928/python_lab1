#Local and Global variables


```python
a=10     #global var
b=20
print(a,b)
def func():
  c=100   #local
  print (c)
  print (a,b)
func()
```

```python
a=10
b=20
print(a,b)
def func():
  c=100
  print (c)
  print (a,b)
a=120
func()
print(a,b)
```

```python
a=10
b=20
print(a,b)
def func():
  c=100
  a=120
  print (c)
  print (a,b)

func()
print(a,b)
```

```python
a=10
b=20
print(a,b)
def func():
  c=100
  a= 300
  print (c)
  print (a,b)
a=120
func()
print(a,b)
```

```python
a=10
b=20
print(a,b)
def func():
  c=100
  global a   #to change the value of global variable
  a=110
  print (c)
  print (a,b)
func()
print(a,b)
```
