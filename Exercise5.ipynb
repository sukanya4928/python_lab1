```python
#program that count the number of vowels in a string
a = str(input("Enter a string: "))
b = a.lower()
vowels = "aieou"
count_v = 0
count_c = 0

for ch in b:
    if ch.isalpha():
        if ch in vowels:
            count_v += 1
        else:
            count_c += 1

print("no of vowels are", count_v)
print("no of consonants are", count_c)
```
