```python
#program that prints reverse of the number using functions
def rev(number):
  r=0
  while(number!=0):
    rem=number%10
    r=r*10+rem
    number=number//10
  return r
x=rev(1234)
print(x)
```

```python
#program that count the no. of vowels and consonants in a string
a=str(input())
b=a.lower()
vowels="aeiou"
count_v=0
count_c=0
for ch in b:
  if ch.isalpha():
    if ch in vowels:
      count_v+=1
    else:
      count_c+=1
print("vowels=",count_v)
print("consonants=",count_c)
```
